Cleaning Data

```python
# Find Missing Values #
Missing_blank_values=pd.DataFrame({
    'Missing_records':df.isna().sum()
})
Missing_blank_values
```

```python
# Check Unique Values (Missing Values)[Categorical]

columns=[
    'Teacher_Quality',
    'Parental_Education_Level',
    'Distance_from_Home'
]
for col in columns:
  print(col)
  print(df[col].value_counts(dropna=False))
  print()
```

```python
# Handle Missing Values Using Mode

for col in columns:
  df[col]=df[col].fillna(df[col].mode()[0])
```

```python
# Verify Missing Values

Missing_values_status=pd.DataFrame({
    'Missing_records':df.isnull().sum()
})
Missing_values_status
```

```python
# Find Duplicated Values

df.duplicated().sum()
```

```python
# Business Scenario-1
# What is the average exam score of students, and how does the overall academic performance look?

plt.figure(figsize=(10,5))
sns.histplot(data=df,x='Exam_Score',bins=15)
plt.title('Distribution of Student Exam Score')
plt.xlabel('Exam Score')
plt.ylabel('Number of Students')
plt.show()

average_exam_score= df['Exam_Score'].mean()
print('Average Score:',average_exam_score)
```

Insight:

  The Histogram shows the distribution of students' exam scores. The average exam score provides an overall measure of academic performance. The distribution also helps identify the score ranges where most students are concentrated.

Business Recommendation:

  Schools can identify students who may need additional academic support.

  Teachers can provide extra learning opportunities based on students' performance.

  Schools can monitor exam-score distributions to evaluate overall academic progress.

```python
# Business Scenario-2
# Which students previous academic performance is associated with higher current exam scores?

plt.figure(figsize=(8,5))
df['Performance']=pd.cut(
    df['Exam_Score'],
    bins=3,
    labels=['Low','Medium','High']
)
sns.scatterplot(
    data=df,
    x='Previous_Scores',
    y='Exam_Score',
    hue='Performance',
    palette='viridis',
    alpha=0.7
)
plt.title('Previous Score Vs Current Exam Score')
plt.xlabel('Previous Score')
plt.ylabel('Current Exam Scores')
plt.legend(title='Current Performance')
plt.show()
```

Insights

  The scatter plot illustrates the relationship between students' previous academic scores and their current exam scores. It helps identify whether students with higher previous scores tend to achieve better results in their current examinations.

Business Recommendations

  Schools should use previous academic records to identify students who may need additional academic support.

  Teachers should develop targeted learning strategies to improve students' academic performance.

  Regular academic assessments should be conducted to monitor student progress and identify areas requiring improvement.

```python
# Business Scenario-3
# What percentage of students fall into different exam performance groups?

plt.figure(figsize=(8,5))
df['Performance']=pd.cut(
    df['Exam_Score'],
    bins=3,
    labels=['Low','High','Medium'],
    include_lowest=True
)
performance_percentage=(
    df['Performance']
    .value_counts(normalize=True)
    .reindex(['Low','Medium','High'])*100
)
plt.figure(figsize=(4,6))
ax=sns.barplot(
    x=performance_percentage.index,
    y=performance_percentage.values,
    hue=performance_percentage.index,
    palette='viridis',
    legend=False
)
for container in ax.containers:
    ax.bar_label(
        container,
        label_type='edge',
        fmt='%.1f%%',
        padding=5,
        color='black',
        fontsize=10
    )
plt.title('Percentage of Students by Exam Performance')
plt.xlabel('Performance Groups')
plt.ylabel('Percentage of Students')
plt.ylim(0,performance_percentage.max()+10)
plt.show()
```

Insights

 The bar plot illustrates the percentage distribution of students across Low, Medium and High exam performance groups. It helps identify which performance group contains the largest and smallest proportion of students and provides an overview of the overall academic performance distribution.

Business Recommendations

 Schools should prioritize academic support programmes for students belonging to the Low Performance group.

 Teachers should introduce targeted learning activities to help Medium Performance students improve their academic results.

 Schools should regularly monitor performance-group percentages to evaluate the effectiveness of academic improvement programmes.

```python
# Business Scenario-4
# How do study hours relate to students' exam scores?

plt.figure(figsize=(10,5))
sns.regplot(
    data=df,
    x='Hours_Studied',
    y='Exam_Score',
    scatter_kws={'color':'steelblue','s':10},
    line_kws={'color':'red','linewidth':2}
)
r2=df['Hours_Studied'].corr(df['Exam_Score'])**2
plt.xticks(range(0,21,5))
plt.yticks(range(20,101,10))
plt.title('Study Hours Vs Exam Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.grid(alpha=0.3)
plt.show()
```

Insights

 The regression plot illustrates the relationship between students' study hours and their exam scores. The trend line helps identify whether increased study hours are associated with higher exam scores, while the distribution of points shows variations in individual student performance.

Business Recommendations

 Schools should encourage students to develop consistent and effective study habits.

 Teachers should provide personalized study plans to students who require additional academic support.

 Schools should monitor study patterns and academic performance to identify effective learning strategies and improve student outcomes.

```python
# Business Scenario-5
# How does student attendance relate to academic performance?

plt.figure(figsize=(10,5))
df['Attendance'] = pd.cut(
    df['Attendance'],
    bins=[59, 70, 80, 90, 100],
    labels=['60-70%', '71-80%', '81-90%', '91-100%']
)
sns.boxplot(
    data=df,
    x='Attendance',
    y='Exam_Score'
)
plt.title('Exam Score Distribution by Attendance Group')
plt.xlabel('Attendance Group')
plt.ylabel('Exam Score')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

Insights

 The box plot illustrates the distribution of students' exam scores across different attendance groups. It helps compare the median scores, score variations and potential outliers to understand how attendance levels are associated with academic performance.

Business Recommendations

 Schools should encourage regular attendance through student engagement and attendance-monitoring programmes.

 Teachers should identify students with low attendance and provide appropriate academic support to help them keep up with lessons.

 Schools should regularly analyse attendance patterns and exam results to develop strategies for improving student participation and academic performance.

```python
# Business Scenario-6
# Which study-hour group has the highest average exam score?

df['Hours_Studied_Group']=pd.cut(
    df['Hours_Studied'],
    bins=[0,5,10,15,20],
    labels=['0-5','6-10','11-15','16-20'],
    include_lowest=True
)
study_analysis=df.groupby(
    'Hours_Studied_Group',
    observed=False
)['Exam_Score'].mean()
print(study_analysis)
print('Highest Average Score:',study_analysis.idxmax())
plt.figure(figsize=(4,6))
ax=sns.barplot(
    x=study_analysis.index,
    y=study_analysis.values,
    hue=study_analysis.index,
    palette='viridis',
    legend=False
)
for container in ax.containers:
  ax.bar_label(
      container,
      label_type='edge',
      color='black',
      fontsize=10
  )
plt.title('Average Exam Score by Study Hours')
plt.xlabel('Study Hours Group')
plt.ylabel('Average Exam Score')
plt.show()
```

Insights

 The bar plot compares the average exam scores across different study-hour groups. It identifies which study-hour group achieves the highest average score and helps understand how study duration is associated with students' academic performance.

Business Recommendations

 Schools should encourage students to develop consistent study schedules based on effective learning practices.

 Teachers should identify study-hour groups with lower average scores and provide additional academic guidance.

 Educational institutions should promote balanced study routines and regularly evaluate students' academic progress.
